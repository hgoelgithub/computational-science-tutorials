```
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Dummy dataset
data = pd.DataFrame({
    "MW":[100,200,300],
    "LogP":[1,3,5],
    "Toxic":[0,0,1]
})

X = data[["MW","LogP"]]
y = data["Toxic"]

model = RandomForestClassifier()
model.fit(X,y)

print(model.predict([[250,4]]))
```
