# 1. 🐼 Pandas — Data Analysis
> **Python + Pandas = Data Analysis**

Pandas is the backbone of data science in Python. It gives you a `DataFrame` —
think of it as a supercharged Excel spreadsheet that you can manipulate with code.

**Key concepts:** DataFrame, Series, groupby, merge, pivot_table, missing data

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

# ── Load the Titanic dataset (classic, built into seaborn) ───────────────────
import urllib.request, io
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
try:
    df = pd.read_csv(url)
except:
    # Offline fallback — create realistic synthetic Titanic data
    np.random.seed(42)
    n = 891
    df = pd.DataFrame({
        "PassengerId": range(1, n+1),
        "Survived":  np.random.choice([0,1], n, p=[0.62, 0.38]),
        "Pclass":    np.random.choice([1,2,3], n, p=[0.24,0.21,0.55]),
        "Name":      [f"Person_{i}" for i in range(n)],
        "Sex":       np.random.choice(["male","female"], n, p=[0.65,0.35]),
        "Age":       np.random.normal(30, 14, n).clip(1, 80).round(1),
        "Fare":      np.random.lognormal(3.2, 1.0, n).round(2),
        "Embarked":  np.random.choice(["S","C","Q"], n, p=[0.72,0.19,0.09]),
    })
    df.loc[np.random.choice(df.index, 177, replace=False), "Age"] = np.nan

print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print()
print("── First 5 rows ──")
print(df.head())
````

````python
# ── Basic statistics ──────────────────────────────────────────────────────────
print("── Summary statistics ──")
print(df[["Age","Fare","Survived"]].describe().round(2))

print()
print("── Missing values ──")
print(df.isnull().sum())

# ── Data cleaning ─────────────────────────────────────────────────────────────
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Embarked"].fillna("S", inplace=True)
df["FareCategory"] = pd.cut(df["Fare"], bins=[0,10,30,100,600],
                             labels=["Low","Medium","High","Premium"])
print()
print("── After cleaning — Age nulls:", df["Age"].isnull().sum())
````

````python
# ── GroupBy analysis ──────────────────────────────────────────────────────────
print("── Survival rate by Class and Sex ──")
survival = df.groupby(["Pclass","Sex"])["Survived"].agg(["mean","sum","count"])
survival.columns = ["Survival_Rate","Survivors","Total"]
survival["Survival_Rate"] = survival["Survival_Rate"].round(3)
print(survival)

print()
print("── Average fare per class ──")
print(df.groupby("Pclass")["Fare"].mean().round(2))

# ── Visualisation ─────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

df.groupby("Sex")["Survived"].mean().plot(kind="bar", ax=axes[0],
    color=["#3498DB","#E74C3C"], edgecolor="white", alpha=0.85)
axes[0].set_title("Survival Rate by Sex", fontweight="bold")
axes[0].set_ylabel("Survival Rate"); axes[0].set_xticklabels(["Female","Male"], rotation=0)

df.groupby("Pclass")["Survived"].mean().plot(kind="bar", ax=axes[1],
    color=["#27AE60","#F1C40F","#E67E22"], edgecolor="white", alpha=0.85)
axes[1].set_title("Survival Rate by Class", fontweight="bold")
axes[1].set_ylabel("Survival Rate"); axes[1].set_xticklabels(["1st","2nd","3rd"], rotation=0)

df["Age"].hist(bins=30, ax=axes[2], color="#8E44AD", edgecolor="white", alpha=0.85)
axes[2].set_title("Age Distribution", fontweight="bold")
axes[2].set_xlabel("Age"); axes[2].set_ylabel("Count")

plt.suptitle("Pandas — Titanic Dataset Analysis", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("pandas_titanic.png", dpi=120, bbox_inches="tight")
plt.show()
print("Key insight: Women had 3x higher survival rate than men (class 1 women: ~97%)")
````

## Deep Dive: Pandas

### Why Pandas?
Before Pandas, data analysis meant painful CSV loops. Pandas gives you a **DataFrame** — a 2D table with labelled rows and columns, like Excel but programmable and far more powerful.

### Core Data Structures
- **Series**: 1D labelled array (one column)
- **DataFrame**: 2D labelled table (the spreadsheet)
- **Index**: Row/column labels (dates, IDs, strings)

### The Titanic Dataset
891 passengers, real 1912 survival data. Mix of numeric, categorical, and missing data — perfect for learning. Key columns: Survived (0/1), Pclass (1/2/3), Sex, Age, Fare, Embarked.

### Essential Operations
```python
df.head(5)           # first 5 rows
df.describe()        # count/mean/std/min/max for each column
df.isnull().sum()    # count missing values per column
df.fillna(value)     # replace NaN with a value
df.groupby(col).agg() # split-apply-combine
df.merge(other, on="id") # SQL-style JOIN
df.pivot_table(values, index, columns, aggfunc)
```

### Handling Missing Data
177 of 891 Age values are missing (20%). Strategy used here: **fill with median** — robust to outliers. Alternative: `df.dropna(subset=["Age"])` removes rows with missing Age.

### Key Insight
Women had 74% survival rate vs 19% for men — "women and children first" is clearly visible in the data. 1st class had 63% survival vs 24% for 3rd class. Data tells the story.


## ✅ Key Takeaways — 🐼 Pandas

1. Pandas DataFrames make data manipulation intuitive and powerful
2. Always check for missing values before analysis (df.isnull().sum())
3. groupby() + agg() is the workhorse for exploratory data analysis
4. The Titanic dataset reveals clear patterns: gender > class in survival
