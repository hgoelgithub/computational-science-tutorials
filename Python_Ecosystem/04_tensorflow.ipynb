# 4. 🧠 TensorFlow / Keras — Deep Learning
> **Python + TensorFlow = Deep Learning**

TensorFlow (Google) is the production-grade deep learning framework.
Keras (its high-level API) makes building networks as easy as stacking LEGO bricks.

**Key concepts:** Sequential model, layers, compile, fit, evaluate, callbacks

````python
# ── TensorFlow / Keras walkthrough ───────────────────────────────────────────
try:
    import tensorflow as tf
    from tensorflow import keras
    from tensorflow.keras import layers
    TF_OK = True
    print(f"TensorFlow version: {tf.__version__}")
except ImportError:
    TF_OK = False
    print("TensorFlow not installed — pip install tensorflow")
    print("Showing code structure only.")

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

iris = load_iris()
X, y = iris.data.astype("float32"), iris.target
sc = StandardScaler()
X = sc.fit_transform(X)

# One-hot encode labels for Keras
if TF_OK:
    y_cat = keras.utils.to_categorical(y, 3)
    X_train, X_test, y_train, y_test = train_test_split(X, y_cat, test_size=0.2, random_state=42)

    # ── Build the model ───────────────────────────────────────────────────────
    model = keras.Sequential([
        layers.Input(shape=(4,)),
        layers.Dense(64, activation="relu"),      # fully connected layer
        layers.BatchNormalization(),               # normalise activations
        layers.Dropout(0.3),                       # prevent overfitting
        layers.Dense(32, activation="relu"),
        layers.Dense(3,  activation="softmax"),    # 3 output probabilities
    ], name="IrisClassifier")

    model.compile(
        optimizer="adam",
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    model.summary()
else:
    print("""
    model = keras.Sequential([
        layers.Input(shape=(4,)),
        layers.Dense(64, activation="relu"),
        layers.BatchNormalization(),
        layers.Dropout(0.3),
        layers.Dense(32, activation="relu"),
        layers.Dense(3,  activation="softmax"),
    ])
    model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
    """)
````

````python
if TF_OK:
    # ── Train with callbacks ─────────────────────────────────────────────────
    callbacks = [
        keras.callbacks.EarlyStopping(patience=20, restore_best_weights=True),
        keras.callbacks.ReduceLROnPlateau(factor=0.5, patience=10, verbose=0)
    ]

    history = model.fit(
        X_train, y_train,
        epochs=200,
        batch_size=16,
        validation_split=0.2,
        callbacks=callbacks,
        verbose=0
    )

    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    print(f"Test accuracy: {acc:.3f}")

    # ── Plot training history ─────────────────────────────────────────────────
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(history.history["loss"],     label="Train", color="#E74C3C", lw=2)
    axes[0].plot(history.history["val_loss"], label="Val",   color="#3498DB", lw=2)
    axes[0].set_title("Loss", fontweight="bold"); axes[0].legend(); axes[0].grid(True,alpha=0.3)

    axes[1].plot(history.history["accuracy"],     label="Train", color="#27AE60", lw=2)
    axes[1].plot(history.history["val_accuracy"], label="Val",   color="#F1C40F", lw=2)
    axes[1].set_title("Accuracy", fontweight="bold"); axes[1].legend(); axes[1].grid(True,alpha=0.3)

    plt.suptitle("TensorFlow/Keras — Training History", fontsize=13, fontweight="bold")
    plt.tight_layout()
    plt.savefig("tensorflow_history.png", dpi=120, bbox_inches="tight")
    plt.show()
else:
    print("Run after: pip install tensorflow")
    print("Key advantage over PyTorch: model.fit() handles the entire training loop")
````

## Deep Dive: TensorFlow/Keras

### Keras vs TensorFlow
- **TensorFlow**: low-level ops, tensors, sessions (complex)
- **Keras**: high-level API on top (simple)
- Since TF 2.0 (2019): fully merged. Use `tf.keras` or `from tensorflow import keras`

### Sequential vs Functional API
```python
# Sequential: linear stack (most common)
model = keras.Sequential([Dense(64, activation="relu"), Dense(3)])

# Functional: for complex architectures (multiple inputs, skip connections)
inputs = keras.Input(shape=(4,))
x = Dense(64, activation="relu")(inputs)
outputs = Dense(3, activation="softmax")(x)
model = keras.Model(inputs, outputs)
```

### Key Layers Explained
- **Dense(64, activation="relu")**: fully connected, 64 neurons, ReLU activation
- **BatchNormalization()**: normalises layer outputs — faster training, less sensitive to learning rate
- **Dropout(0.3)**: zeros 30% of neurons randomly — regularisation against overfitting

### compile() Unpacked
```python
model.compile(
    optimizer="adam",                  # Adam: adaptive learning rates, usually best default
    loss="categorical_crossentropy",   # multi-class classification loss
    metrics=["accuracy"]               # what to show during training
)
```

### Callbacks
EarlyStopping(patience=20) monitors val_loss. If it hasn't improved for 20 epochs, training stops and the best weights are restored. Prevents overfitting and wasted compute.


## ✅ Key Takeaways — 🧠 TensorFlow/Keras

1. Keras makes neural networks as simple as stacking LEGO layers
2. BatchNormalization + Dropout are your best anti-overfitting tools
3. EarlyStopping automatically stops training when validation plateaus
4. model.fit() handles the entire training loop in one line
