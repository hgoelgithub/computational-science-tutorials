# 6. 🔢 NumPy — Scientific Computing
> **Python + NumPy = Scientific Computing**

NumPy is the foundation of all scientific Python. Provides fast N-dimensional arrays
(ndarray) and mathematical operations that run at C speed.

**Key concepts:** ndarray, broadcasting, vectorisation, linear algebra, FFT, random

````python
import numpy as np
import matplotlib.pyplot as plt

# ── Array creation and operations ─────────────────────────────────────────────
print("── Array basics ──")
a = np.array([1, 2, 3, 4, 5])
b = np.arange(1, 6)
print(f"a = {a}")
print(f"a² = {a**2}")
print(f"√a = {np.sqrt(a).round(3)}")
print(f"sum={a.sum()}, mean={a.mean()}, std={a.std():.3f}")

# ── Broadcasting: operate on arrays of different shapes ──────────────────────
print("\n── Broadcasting ──")
matrix = np.ones((3, 4))
row    = np.array([1, 2, 3, 4])
result = matrix + row   # row is "broadcast" across all 3 rows
print(f"matrix (3×4) + row (4,):\n{result}")

# ── Real example: Signal processing ──────────────────────────────────────────
print("\n── Signal processing (FFT) ──")
t  = np.linspace(0, 1, 1000)       # 1 second, 1000 samples
s1 = np.sin(2 * np.pi * 5  * t)    # 5 Hz sine wave
s2 = np.sin(2 * np.pi * 20 * t)    # 20 Hz sine wave
noise  = 0.3 * np.random.randn(len(t))
signal = s1 + s2 + noise

# Fast Fourier Transform
fft_vals = np.abs(np.fft.rfft(signal))
freqs    = np.fft.rfftfreq(len(t), d=1/1000)
print(f"Signal: 5 Hz + 20 Hz + noise")
print(f"Top frequency peaks: {freqs[np.argsort(fft_vals)[-5:]][::-1].round(1)} Hz")
````

````python
# ── Linear Algebra: fundamental to all ML ────────────────────────────────────
print("── Linear Algebra ──")
A = np.array([[2, 1], [5, 3]])
b = np.array([4, 7])
x = np.linalg.solve(A, b)     # solve Ax = b
print(f"Solving Ax=b: x = {x}")

eigenvalues, eigenvectors = np.linalg.eig(A)
print(f"Eigenvalues: {eigenvalues.round(3)}")

# ── Visualisation ─────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Plot 1: Signal in time domain
axes[0].plot(t[:200], signal[:200], color="#3498DB", lw=1, alpha=0.8)
axes[0].set_title("Noisy Signal (time domain)", fontweight="bold")
axes[0].set_xlabel("Time (s)"); axes[0].set_ylabel("Amplitude")
axes[0].grid(True, alpha=0.3)

# Plot 2: Frequency spectrum (FFT)
axes[1].plot(freqs[:100], fft_vals[:100], color="#E74C3C", lw=2)
axes[1].axvline(5,  c="#27AE60", ls="--", lw=2, label="5 Hz")
axes[1].axvline(20, c="#F1C40F", ls="--", lw=2, label="20 Hz")
axes[1].set_title("Frequency Spectrum (FFT)", fontweight="bold")
axes[1].set_xlabel("Frequency (Hz)"); axes[1].set_ylabel("|FFT|")
axes[1].legend(); axes[1].grid(True, alpha=0.3)

# Plot 3: Broadcasting / vectorisation speed demo
import time
n = 1_000_000
arr = np.random.randn(n)
t0 = time.time()
_ = arr ** 2 + 2*arr           # NumPy vectorised
np_time = time.time() - t0

list_arr = list(arr)
t0 = time.time()
_ = [x**2 + 2*x for x in list_arr]    # pure Python loop
py_time = time.time() - t0

axes[2].bar(["Python loop", "NumPy vectorised"],
            [py_time*1000, np_time*1000],
            color=["#E74C3C","#27AE60"], alpha=0.85, edgecolor="white")
axes[2].set_title(f"Speed: NumPy {py_time/np_time:.0f}× faster!", fontweight="bold")
axes[2].set_ylabel("Time (ms)")
for i, v in enumerate([py_time*1000, np_time*1000]):
    axes[2].text(i, v+0.1, f"{v:.1f}ms", ha="center", fontweight="bold")

plt.suptitle("NumPy — Scientific Computing", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("numpy_demo.png", dpi=120, bbox_inches="tight")
plt.show()
````

## Deep Dive: NumPy

### Why NumPy is 50-200x Faster
Python lists store objects with type information (28+ bytes each, type-checked at runtime). NumPy stores numbers in contiguous C arrays (4-8 bytes each) and calls optimised BLAS/LAPACK routines with CPU SIMD vectorisation.

### Broadcasting — Powerful Implicit Loops
```python
matrix = np.ones((3, 4))      # shape (3, 4)
row    = np.array([1, 2, 3, 4])  # shape (4,)
result = matrix + row          # row broadcast to (3,4) — no loop needed
```
Rule: dimensions compatible if equal OR one of them is 1. NumPy stretches the smaller array mentally.

### The FFT for Signal Processing
FFT (Fast Fourier Transform) decomposes any signal into its frequencies:
- Time domain: you see the noisy waveform
- Frequency domain (after FFT): you see sharp peaks at exactly 5 Hz and 20 Hz

Used in: audio processing, MRI reconstruction, spectroscopy, molecular dynamics.

### Linear Algebra Essentials
```python
np.linalg.solve(A, b)  # solve Ax = b (linear systems)
np.linalg.eig(A)        # eigenvalues and eigenvectors
np.linalg.svd(A)        # singular value decomposition (basis of PCA)
A @ B                    # matrix multiplication (Python 3.5+)
```

### Vectorisation Habit
Always ask: "Can I replace this Python loop with a NumPy operation?" If yes — do it. The speedup is almost always 50x or more.


## ✅ Key Takeaways — 🔢 NumPy

1. NumPy arrays are 50-200x faster than Python lists for numerical work
2. Broadcasting eliminates nearly all explicit loops
3. FFT is the key algorithm for frequency-domain analysis
4. Always think: can I replace this loop with a vectorised operation?
