# 15. 🌐 Selenium — Web Automation
> **Python + Selenium = Web Automation**

Selenium controls a real browser programmatically. Unlike BeautifulSoup (static HTML),
Selenium handles **JavaScript-rendered pages**, login forms, and dynamic content.

**Key concepts:** WebDriver, find_element, click, send_keys, wait, headless browser

````python
# ── Selenium code patterns ────────────────────────────────────────────────────
print("Selenium Web Automation Patterns")
print("(Requires: pip install selenium webdriver-manager + Chrome installed)")
print()

selenium_code = """
# ── Setup ─────────────────────────────────────────────────────────────────────
pip install selenium webdriver-manager

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time, pandas as pd

# ── Launch Chrome (headless = no visible window) ──────────────────────────────
options = Options()
options.add_argument("--headless")           # run without opening a browser window
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(ChromeDriverManager().install(), options=options)
wait   = WebDriverWait(driver, 10)          # wait up to 10 seconds for elements

# ══════════════════════════════════════════════════════════════════════════════
# PATTERN 1: Google search automation
# ══════════════════════════════════════════════════════════════════════════════
driver.get("https://www.google.com")
search_box = driver.find_element(By.NAME, "q")    # find the search input
search_box.send_keys("PFAS toxicology 2024")       # type into it
search_box.send_keys(Keys.RETURN)                  # press Enter

# Wait for results and extract titles
time.sleep(2)
results = driver.find_elements(By.CSS_SELECTOR, "h3")
for i, r in enumerate(results[:5]):
    print(f"{i+1}. {r.text}")

# ══════════════════════════════════════════════════════════════════════════════
# PATTERN 2: Log into a web app, fill a form
# ══════════════════════════════════════════════════════════════════════════════
driver.get("https://example-chemicals.com/login")
driver.find_element(By.ID, "username").send_keys("myuser")
driver.find_element(By.ID, "password").send_keys("mypassword")
driver.find_element(By.XPATH, "//button[@type='submit']").click()

# ══════════════════════════════════════════════════════════════════════════════
# PATTERN 3: Wait for dynamic content (JavaScript-rendered)
# ══════════════════════════════════════════════════════════════════════════════
driver.get("https://pubchem.ncbi.nlm.nih.gov/compound/Aspirin")
table = wait.until(
    EC.presence_of_element_located((By.CLASS_NAME, "data-table"))
)
print("Table loaded:", table.text[:200])

# ══════════════════════════════════════════════════════════════════════════════
# PATTERN 4: Screenshot & save data
# ══════════════════════════════════════════════════════════════════════════════
driver.save_screenshot("page_screenshot.png")      # save what browser sees
rows = driver.find_elements(By.TAG_NAME, "tr")     # extract table rows
data = [[td.text for td in row.find_elements(By.TAG_NAME, "td")] for row in rows]
df   = pd.DataFrame(data)
df.to_csv("scraped_data.csv", index=False)

driver.quit()   # ALWAYS close the browser when done!
"""

print(selenium_code)

# ── When to use Selenium vs BeautifulSoup ─────────────────────────────────────
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Comparison chart
categories = ["JavaScript\nContent","Login\nForms","Dynamic\nData","Screenshots",
               "Speed","Simplicity"]
beautiful_scores = [1, 2, 1, 1, 9, 9]
selenium_scores  = [10,10, 9,10, 3, 5]

x = np.arange(len(categories))
w = 0.35
axes[0].bar(x-w/2, beautiful_scores, w, label="BeautifulSoup", color="#27AE60", alpha=0.85, edgecolor="white")
axes[0].bar(x+w/2, selenium_scores,  w, label="Selenium",      color="#E74C3C", alpha=0.85, edgecolor="white")
axes[0].set_xticks(x); axes[0].set_xticklabels(categories, fontsize=8)
axes[0].set_ylabel("Score (0-10)"); axes[0].set_title("Selenium vs BeautifulSoup\nCapability Comparison", fontweight="bold")
axes[0].legend(); axes[0].grid(True, alpha=0.3, axis="y")

# Selenium workflow diagram
axes[1].set_xlim(0,10); axes[1].set_ylim(0,10); axes[1].axis("off")
axes[1].set_facecolor("#F8F9FA")

workflow = [
    (5, 8.5, "Python Script\n(your code)", "#3498DB"),
    (5, 6.5, "Selenium\nWebDriver", "#8E44AD"),
    (5, 4.5, "Chrome Browser\n(real or headless)", "#27AE60"),
    (5, 2.5, "Website\n(JavaScript OK!)", "#E74C3C"),
]
for x, y, label, col in workflow:
    axes[1].add_patch(mpatches.FancyBboxPatch((x-2.0,y-0.6),4.0,1.2,
        boxstyle="round,pad=0.1",facecolor=col,edgecolor="white",alpha=0.85,lw=2))
    axes[1].text(x, y, label, ha="center", va="center", color="white",
                  fontweight="bold", fontsize=9)

for i in range(len(workflow)-1):
    y1, y2 = workflow[i][1]-0.6, workflow[i+1][1]+0.6
    axes[1].annotate("", xy=(5,y2), xytext=(5,y1),
                      arrowprops=dict(arrowstyle="->",color="#1F2937",lw=2))

axes[1].text(5, 9.5, "Selenium Automation Stack", ha="center",
              fontsize=11, fontweight="bold")
plt.suptitle("Selenium — Web Automation", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("selenium_automation.png", dpi=120, bbox_inches="tight")
plt.show()
````

## Deep Dive: Selenium

### Why Selenium for Dynamic Websites
BeautifulSoup downloads HTML and parses it. But modern websites (LinkedIn, Twitter, PubChem) render content with JavaScript AFTER the page loads. BeautifulSoup sees the empty shell; Selenium waits for the browser to execute JavaScript and sees the full rendered content.

### Finding Elements — Best Strategy
Order of reliability (most to least stable):
1. By.ID — most unique, fastest lookup
2. By.CSS_SELECTOR — powerful, readable, like stylesheets
3. By.XPATH — most powerful, can navigate backwards up the tree
4. By.TEXT — fragile, breaks if text changes

### The WebDriverWait Pattern
```python
wait = WebDriverWait(driver, 10)

# Wait until element exists in DOM
element = wait.until(EC.presence_of_element_located((By.ID, "result")))

# Wait until element is visible and clickable
button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, ".submit-btn")))
```
Never use `time.sleep(3)` — it's slow when the page loads fast and breaks when it loads slowly.

### Headless Mode
Adding `--headless` runs Chrome as a background process with no window. Essential for:
- Running in CI/CD pipelines (no display server)
- Running on remote servers
- Scheduled automated jobs

### Common Pitfalls
- StaleElementReferenceException: page changed after you found the element — re-find it
- ElementNotInteractableException: element exists but is hidden — scroll to it or wait for visibility
- TimeoutException: element never appeared — check if you need to log in first


## ✅ Key Takeaways — 🌐 Selenium

1. Selenium runs a real browser — JavaScript renders, logins work, SPAs are supported
2. WebDriverWait is mandatory — never use time.sleep() for page loads
3. Always call driver.quit() when done — leaked Chrome processes fill RAM fast
4. --headless mode runs Chrome invisibly on servers and in CI/CD
