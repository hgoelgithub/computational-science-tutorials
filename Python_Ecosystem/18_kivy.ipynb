# 18. 📱 Kivy — Desktop & Mobile App Development
> **Python + Kivy = Desktop App Development**

Kivy builds cross-platform GUI apps that run on Windows, Mac, Linux, iOS, and Android.
One codebase → deploys everywhere.

**Key concepts:** App, widgets, layout, events, kv language, multi-touch

````python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

print("Kivy cross-platform app code:")
print("pip install kivy[full]\n")

kivy_code = """
from kivy.app import App
from kivy.uix.boxlayout import BoxLayout
from kivy.uix.label import Label
from kivy.uix.button import Button
from kivy.uix.slider import Slider
from kivy.uix.textinput import TextInput
from kivy.graphics import Color, Rectangle
import json

class ToxicityApp(App):
    """Chemical toxicity calculator — runs on Windows, Mac, Linux, Android!"""

    def build(self):
        # ── Main layout: vertical stack ──────────────────────────────────────
        layout = BoxLayout(orientation="vertical", padding=20, spacing=10)

        layout.add_widget(Label(text="Chemical Toxicity Estimator", font_size=22,
                                  bold=True, color=(0.2, 0.6, 1, 1)))

        # ── Input: chemical name ──────────────────────────────────────────────
        self.name_input = TextInput(hint_text="Enter chemical name (e.g. aspirin)",
                                      multiline=False, font_size=16)
        layout.add_widget(self.name_input)

        # ── Sliders for MW and logP ────────────────────────────────────────────
        for label, attr, min_v, max_v, val in [
            ("Molecular Weight:", "mw_slider", 50, 800, 200),
            ("logP (lipophilicity):", "logp_slider", -3, 7, 2),
        ]:
            layout.add_widget(Label(text=label, font_size=14, color=(0.8,0.8,0.8,1)))
            slider = Slider(min=min_v, max=max_v, value=val, step=0.1)
            slider.bind(value=self.on_slider_change)
            setattr(self, attr, slider)
            layout.add_widget(slider)

        # ── Predict button ────────────────────────────────────────────────────
        btn = Button(text="Predict Toxicity", font_size=16,
                      background_color=(0.2, 0.8, 0.4, 1))
        btn.bind(on_press=self.predict)
        layout.add_widget(btn)

        # ── Result label ──────────────────────────────────────────────────────
        self.result = Label(text="Result will appear here", font_size=14,
                             color=(1, 0.8, 0, 1))
        layout.add_widget(self.result)
        return layout

    def on_slider_change(self, instance, value):
        pass  # could update display live

    def predict(self, instance):
        name = self.name_input.text or "Unknown"
        mw   = self.mw_slider.value
        logp = self.logp_slider.value

        # Simple rule-based prediction
        if logp > 5 or mw > 500:
            risk = "HIGH concern — check further"
        elif logp > 3:
            risk = "MODERATE — monitor bioaccumulation"
        else:
            risk = "LOW concern"

        lip = "Ro5: PASS" if (mw<=500 and logp<=5) else "Ro5: FAIL"
        self.result.text = f"{name}: {risk} | {lip}"

if __name__ == "__main__":
    ToxicityApp().run()
"""
print(kivy_code)

# ── Simulated UI screenshot ───────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(6, 9))
ax.set_facecolor("#1A1A2E"); ax.axis("off")
ax.set_xlim(0, 6); ax.set_ylim(0, 9)

# Phone frame
ax.add_patch(mpatches.FancyBboxPatch((0.3,0.3),5.4,8.4,
    boxstyle="round,pad=0.3",facecolor="#16213E",edgecolor="#0F3460",lw=4))

# App title
ax.text(3, 8.2, "Chemical Toxicity Estimator",
         ha="center", color="#4FC3F7", fontsize=12, fontweight="bold")

# Input field
ax.add_patch(mpatches.FancyBboxPatch((0.6,7.1),4.8,0.6,
    boxstyle="round,pad=0.05",facecolor="#0F3460",edgecolor="#4FC3F7",lw=1))
ax.text(3, 7.42, "aspirin", ha="center", color="white", fontsize=11)

# Sliders
for i, (label, val, total) in enumerate([
    ("Molecular Weight: 180", 180, 800),
    ("logP: 1.2", 1.2, 7),
]):
    y = 6.2 - i*1.3
    ax.text(1, y+0.35, label, color="#E0E0E0", fontsize=9)
    ax.add_patch(mpatches.FancyBboxPatch((0.6,y-0.1),4.8,0.28,
        boxstyle="round,pad=0.02",facecolor="#0F3460",edgecolor="none"))
    frac = val/total if total > 0 else 0.5
    ax.add_patch(mpatches.FancyBboxPatch((0.6,y-0.1),4.8*frac,0.28,
        boxstyle="square,pad=0",facecolor="#4FC3F7",alpha=0.8,edgecolor="none"))

# Button
btn = mpatches.FancyBboxPatch((1.0,2.5),4.0,0.65,
    boxstyle="round,pad=0.1",facecolor="#4CAF50",edgecolor="none")
ax.add_patch(btn)
ax.text(3, 2.82, "Predict Toxicity", ha="center", color="white",
         fontsize=11, fontweight="bold")

# Result
ax.add_patch(mpatches.FancyBboxPatch((0.6,1.2),4.8,0.9,
    boxstyle="round,pad=0.08",facecolor="#0F3460",edgecolor="#F1C40F",lw=1.5))
ax.text(3, 1.65, "aspirin: LOW concern | Ro5: PASS",
         ha="center", color="#F1C40F", fontsize=9, fontweight="bold")

# Platform logos
ax.text(3, 0.6, "🪟 Windows  🍎 macOS  🐧 Linux  📱 Android  📱 iOS",
         ha="center", color="#9E9E9E", fontsize=7.5)

ax.set_title("Kivy App — runs on ALL platforms", fontweight="bold", color="black")
plt.tight_layout()
plt.savefig("kivy_app.png", dpi=120, bbox_inches="tight")
plt.show()
````

## Deep Dive: Kivy

### The App Lifecycle

```python
        pass
```

### Widget Layout System
Kivy uses layout widgets to position children:
- **BoxLayout**: stack widgets horizontally or vertically
- **GridLayout**: arrange in a grid (rows x cols)
- **FloatLayout**: absolute positioning (x/y as fractions 0-1)
- **AnchorLayout**: anchor children to corners/center
- **ScrollView**: make content scrollable

### Properties and Data Binding
```python
# KV language: text updates automatically when slider moves
Label:
    text: f"MW: {slider_id.value:.1f} Da"  # live binding!
```
This is Kivy's reactive system — no manual update callbacks needed when using .kv files.

### Cross-Platform Notes
On desktop: runs natively via OpenGL. On Android (via Buildozer): compiles to APK. On iOS (via kivy-ios): requires Mac + Xcode. The same Python codebase runs on all platforms.

### When to Choose Kivy
Kivy is ideal when you need touch support, multi-platform from one codebase, and the app logic is in Python. For purely desktop apps, PyQt6 gives a more native look. For quick command-line tools, tkinter is built into Python.


## ✅ Key Takeaways — 📱 Kivy

1. Kivy's one codebase deploys to Windows, Mac, Linux, Android, and iOS
2. The KV language separates UI structure from logic — like HTML/CSS vs Python
3. widget.bind(property=callback) is the event system for all interactions
4. Buildozer handles the entire Android compilation pipeline from Python
